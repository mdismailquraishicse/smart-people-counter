```python
import cv2
from ultralytics import YOLO
```

```python
source = "public.mp4"
model = YOLO("yolov8n.pt")
cap = cv2.VideoCapture(source)

count_up = 0
count_down = 0
track_history = {}
crossed_ids = set()

while True:
    ret, frame = cap.read()
    if not ret:
        break

    frame = cv2.resize(frame, (1280, 620))
    h, w = frame.shape[:2]
    line_x = w//2
    cv2.line(
        frame,
        (line_x, 0),
        (line_x, h),
        (0, 255, 255),
        2
    )
    results = model.track(frame, classes=[0], verbose=False, persist=True)
    boxes = results[0].boxes
    if boxes is not None and boxes.id is not None:
        ids = boxes.id.cpu().numpy().astype(int)
        xywhs = boxes.xywh.cpu().numpy()
        for tid, xywh in zip(ids, xywhs):
            cx = int(xywh[0])
            cy = int(xywh[1])
            cv2.circle(
                frame,
                (cx, cy),
                4,
                (0, 255, 0),
                -1)
            if tid in track_history:
                prev_x = track_history[tid]
                if tid not in crossed_ids:
                    if prev_x < line_x <= cx:
                        count_down += 1
                        crossed_ids.add(tid)
                    elif prev_x > line_x >= cx:
                        count_up += 1
                        crossed_ids.add(tid)
            track_history[tid] = cx
            cv2.putText(frame, f"Going Down: {count_down}", (20, 40),
                        cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 100, 255), 2)
            cv2.putText(frame, f"Going Up:   {count_up}",   (20, 80),
                        cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 100), 2)

    
    cv2.imshow("Counter", frame)

    if cv2.waitKey(1) == ord("q"):
        break
cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import warnings
from ultralytics import YOLO
warnings.filterwarnings("ignore")
```

```python
counts = {
    "up":0,
    "down": 0,
    "left": 0,
    "right": 0
}
# count_up = 0
# count_down = 0
# count_left = 0
# count_right = 0
track_history = {}
crossed_ids = set()
vid_source = "stairs.mp4"

def update_counters(tid:int, track_history:dict, crossed_ids:set(), line, c, counts:dict):
    if tid in track_history:
        prev = track_history[tid]
        print(f"ID={tid}, prev={prev}, current={c}, line={line}")
        if tid not in crossed_ids:
            if prev < line <= c:
                print(f"up crossing detected")
                crossed_ids.add(tid)
                counts["up"] = 1 + counts["up"]
                print(f"COUNTTTTTTTTTTTTTTTT: {counts}")
            elif prev > line >= c:
                print(f"down crossing detected")
                crossed_ids.add(tid)
                counts["down"] = 1 + counts["down"]
                print(f"COUNTTTTTTTTTTTTTTTT DOWN: {counts}")
    track_history[tid] = c


def draw_centres(frame, ids, xywhs, track_history, crossed_ids, line, counts:dict):

    for tid, xywh in zip(ids, xywhs):
        cx = int(xywh[0])
        cy = int(xywh[1])
        # print(f"tid: {tid} xywh: {xywh} cx: {cx} cy: {cy}")
        cv2.circle(frame, (cx, cy), 4, (0,0,224), -1) # 4: radius, -1: fill entire circle
        update_counters(tid = tid, track_history = track_history, crossed_ids = crossed_ids,
                        line = line, c = cy, count_up = count_up, count_down = count_down)

        # update_counters(tid = tid, track_history = track_history, crossed_ids = crossed_ids,
        #                line = line_x, c = cx, counts = counts)
        # break


def get_boxes_params(boxes):
    if boxes is not None and boxes.id is not None:
        ids = boxes.id.cpu().numpy().astype(int)
        xywhs = boxes.xywh.cpu().numpy()
        return ids, xywhs
```

```python
model = YOLO("yolov8n.pt")
cap = cv2.VideoCapture(vid_source)
while True:
    ret, frame = cap.read()
    if not ret:
        break
    frame = cv2.resize(frame, (1280, 620))
    h,w = frame.shape[:2]
    line_y = h // 2
    line_x = w // 2
    results = model.track(frame, persist=True, classes=[0], save=True)
    ids, xywhs = get_boxes_params(results[0].boxes)
    draw_centres(frame=frame, ids=ids, xywhs=xywhs, track_history=track_history,
                 crossed_ids=crossed_ids, line=line_y, count_up=count_up, count_down=count_down)
    # track_history = cy
    cv2.line(frame, (0, line_y), (w, line_y), (0,255, 255))
    cv2.line(frame, (line_x, 0), (line_x, h), (0, 255, 255))
    
    cv2.putText(frame, f"Going Down: {counts['up']}", (20, 40),
                cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 100, 255), 2)
    cv2.putText(frame, f"Going Up:   {counts['down']}",   (20, 80),
                cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 100), 2)

    # break
    annotated = results[0].plot()
    cv2.imshow("TEST", annotated)
    if cv2.waitKey(1) == ord("q"):
        break
cap.release()
cv2.destroyAllWindows()
```
